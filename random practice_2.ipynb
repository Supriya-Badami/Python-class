```python
# Data Cleaning Practice

name = "  JOHN DOE "
email = " JOHN@GMAIL.COM "
city = " bangalore "
course = " data analyst "

name = name.strip().title()
email = email.strip().lower()
city = city.strip().title()
course = course.strip().title()

print("Name:", name)
print("Email:", email)
print("City:", city)
print("Course:", course)
```
