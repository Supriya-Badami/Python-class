```python
useraccount={
    'username':"codingtraining",
    'findname':'code',
    'lastname':"training",
    'dob':"1-02-2005",
    'pwd':'54321'}
print(useraccount)
```

```python
for a,b in useraccount.items():
    print(a)
    print(b)
```

```python
for a,b in useraccount.items():
    print(f"kay:{a}")
    print(f"value:{b}")
```

```python
for a in useraccount.keys():
    print(f"keys:{a}")
```

```python
for a in useraccount.values():
    print(f"values:{a}")
```
